```python
import sys; sys.path.append("..")
import numpy as np, rasterio
from src.dataset import BASES
from src.inference import load_model, predict_prob

model = load_model()
p = sorted((BASES["atlantic"] / "Test" / "image").glob("*.tif"))[0]
with rasterio.open(p) as src:
    img = src.read()
with rasterio.open(BASES["atlantic"] / "Test" / "mask" / p.name) as src:
    true = src.read(1)

pred = predict_prob(model, img) > 0.5
iou = (pred & (true == 1)).sum() / ((pred | (true == 1)).sum())
print(img.dtype, img.shape, "-> IoU on this tile:", round(float(iou), 3))
```
